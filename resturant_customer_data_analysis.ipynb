```python
# ----------------------------TASK 1---------------------------------------
import pandas as pd
customer_data = {
    "ID":[101,102,103,104,105,106],
    "Name":["Ali","Sara","John","Aman","Riya","David"],
    "Age":[22,25,21,27,23,30]
}
df_1= pd.DataFrame(customer_data)

order_data = {
    "ID":[101,102,103,104,105,107],
    "Order":["Pizza","Burger","Pasta","Cold Coffee","Fries","Mojito"]
}
df_2= pd.DataFrame(order_data)

bill_data = {
    "ID":[101,102,103,104,105,106],
    "Bill":[450,220,380,180,150,500]
}
df_3= pd.DataFrame(bill_data)
# print(df_1)
# print(df_2)
# print(df_3)
```

```python
# ------------------------------------TASK 7---------------------------------------
customer_order_3= pd.merge(left=df_1,right=df_2,how='outer',on='ID')
print(customer_order_3)
```
